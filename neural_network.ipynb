```python
dataset = [
    {
        "intent": "greeting",
        "patterns": ["hi", "hello", "hey", "good morning", "good evening"],
        "responses": ["Hello! How can I help you?", "Hi there! What can I do for you?"]
    },
    {
        "intent": "goodbye",
        "patterns": ["bye", "see you", "good night", "talk to you later"],
        "responses": ["Goodbye!", "See you soon!"]
    },
    {
        "intent": "thanks",
        "patterns": ["thank you", "thanks", "thanks a lot"],
        "responses": ["You're welcome!", "Glad to help!"]
    },
    {
        "intent": "creator",
        "patterns": ["who created you?", "who made you?"],
        "responses": ["I was created by Selva Praveen’s chatbot project."]
    },
    {
        "intent": "weather",
        "patterns": ["what is the weather today?", "tell me today's weather", "weather report"],
        "responses": ["Please tell me your city for weather details."]
    },
    {
        "intent": "name",
        "patterns": ["what is your name?", "who are you?"],
        "responses": ["I am your chatbot assistant."]
    },
    {
        "intent": "age",
        "patterns": ["how old are you?", "what is your age?"],
        "responses": ["I don't have an age. I was created recently!"]
    },
        {
        "intent": "work",
        "patterns": ["what are you doing?"],
        "responses": ["Now I'm assisting you"]
    }
]
```

```python
import numpy as np
from keras.models import Sequential
from keras.layers import Dense, Dropout,Embedding,LSTM
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

```

```python
pattern = []
intent = []
response = {}

for item in dataset:
    for p in item['patterns']:
        pattern.append(p)
        intent.append(item['intent'])
    response[item['intent']] = item['responses']
```

```python
unique_intent = sorted(set(intent))
intent_mp = {intent : idx for idx , intent in enumerate(unique_intent)}
y = np.array([intent_mp[i] for i in intent])

tokenizer = Tokenizer()
tokenizer.fit_on_texts(pattern)
x = tokenizer.texts_to_sequences(pattern)
max_len = max(len(x) for x in x)
x = pad_sequences(x, maxlen=max_len, padding='post')
```

```python
model = Sequential([
    Embedding(input_dim=len(tokenizer.word_index)+1, output_dim=50, input_length=max_len),
    LSTM(64, return_sequences=False),
    Dropout(0.5),
    Dense(32, activation='relu'),
    Dense(len(unique_intent), activation='softmax')
])

model.summary()
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(x,y,epochs=100,batch_size=8,verbose=1,validation_split=0.)
model.evaluate(x,y)
```

```python
def chat_response(user_input):
    seq = tokenizer.texts_to_sequences([user_input])
    padded = pad_sequences(seq,maxlen=max_len,padding='post')
    pred = model.predict(padded)
    intent_idx = np.argmax(pred)
    intent_tag = unique_intent[intent_idx]
    responses = response[intent_tag]
    return np.random.choice(responses)

while True:
    msg = input("You : ")
    if msg.lower() == 'exit':
        print("Chatbot : Goodbye!")
        break
    reply = chat_response(msg)
    print("Chatbot :", reply)
```
